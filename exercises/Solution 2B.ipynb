# Solution for exercise 2B: Positioning loops in a splitter

```python
# ex2 1: input list
names = ["Alfa", "Bravo", "Charlie", "Delta", "Echo", "Foxtrot", "Golf", "Hotel", "India", "Juliett", "Kilo" , "Lima" , "Mike", "November", "Oscar", "Papa", "Quebec", "Romeo", "Sierra", "Tango", "Uniform", "Victor", "Whiskey", "Xray", "Yankee", "Zulu"]
```

```python
# ex2 2: get list length
list_length = len(names)
```

```python
# ex2 3: process list
sentences = []
for name in names:
    sentences.append(f"Hello, {name}! There are {list_length} names.")
```

```python
# ex2 4: print list
print(sentences)
```
