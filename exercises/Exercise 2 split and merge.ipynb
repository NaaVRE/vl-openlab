### Controlling the number of splits
You can manage how many parallel processes are created to handle different parts of a task. This ensures that the task is completed efficiently without overwhelming the system's resources. By balancing the number of splits based on the task size and available resources, you optimize performance and resource usage.

### Contents of a workflow component behind a splitter

When a workflow component is placed behind a splitter, all its code will execute once for each split generated. 
Ensure that a component behind a splitter primarily contains a loop to handle the split data efficiently and that code outside the loop is in a workflow component either before the splitter or after the merger.
This avoinds redundant executions and race conditions.
